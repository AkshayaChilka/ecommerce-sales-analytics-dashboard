```python
import pandas as pd
df = pd.read_csv("E-commerce Sales Data.csv")
```

```python
print(df)
```

```python
print(df.isnull().sum())
```

```python
#clean data
df['Courier Status'] = df['Courier Status'].fillna('Cancelled')
print(df['Courier Status'])

df['fulfilled-by'] = df['fulfilled-by'].fillna('Self Ship')
print(df['fulfilled-by'])
```

```python
print(df.isnull().sum())
```

```python
city_counts = df['ship-city'].value_counts()
df = df[df['ship-city'].isin(city_counts[city_counts > 1].index)]
print(df)
```

```python
print(df.isnull().sum())
```

```python
df['Amount'] = df['Amount'].fillna(df['Amount'].median())
print(df['Amount'])
```

```python
print(df.isnull().sum())
```
